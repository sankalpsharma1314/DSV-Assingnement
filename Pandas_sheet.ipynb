```python
#QU4
import pandas as pd

# 1. Create a sample student DataFrame
data = {
    'StudentID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [20, 21, 19, 22],
    'Grade': ['A', 'B', 'A', 'C']
}
student_df = pd.DataFrame(data)
print("--- Full DataFrame ---")
display(student_df)
name_series = student_df['Name']

print("\n--- 'Name' Column (as Series) ---")
display(name_series)
name_df = student_df[['Name']]

print("\n--- 'Name' Column (as DataFrame) ---")
display(name_df)
```
